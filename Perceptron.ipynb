```python
import torch
print(torch.__version__)
```

```python
x=torch.tensor([8.7,8.2,9.1,7.8,8.2,8.6,8.5,6.7,5.4])
mean, std = x.mean(), x.std()
x_n = (x - mean) / std
y=torch.tensor([1,1,1,0,0,1,0,0,0])
y=y.to(torch.float32)
```

```python
w=torch.tensor(1.0,requires_grad=True)
b=torch.tensor(0.0,requires_grad=True)
epochs = 10000
```

```python
for epoch in range(epochs):
  z=w*x_n+b
  loss=torch.binary_cross_entropy_with_logits(z,y).mean()
  loss.backward()
  with torch.no_grad():
    w-=0.01*w.grad
    b-=0.01*b.grad
  w.grad.zero_()
  b.grad.zero_()
```

```python
x1=(float(input("Enter your CGPA: "))- mean)/std
with torch.no_grad():
  z1=w*x1+b
  y1_pred=torch.sigmoid(z1)
print("The chance of placement: ",y1_pred)
```
